```python
import glob, re
import pandas as pd
import os

pattern = f"/home/michaelgajda/cal_Lab_Work/film/filmScans/filmAlanineConventional_0717/dose_statistics_results/histograms/dose_statistics_results/gaussian_fits_film_*.csv"
files = glob.glob(pattern)
print(f"Found {len(files)} files with pattern: {pattern}")

rows = []
for path in files:
    m = re.search(r"film_(\d+)", path)
    if m is None:
        continue
    num = int(m.group(1))
    df = pd.read_csv(path)
    d_r = df[df["channel"] == "D_r"]
    if d_r.empty:
        continue
    d_r = d_r.iloc[0]
    rows.append({"film": num, "mu": d_r["mu"], "sigma": d_r["sigma"]})

if not rows:
    raise ValueError(f"No matching data found. Pattern: {pattern}")

result = pd.DataFrame(rows).sort_values("film").reset_index(drop=True)
```

```python
result.to_csv('/home/michaelgajda/cal_Lab_Work/film/filmScans/filmAlanineConventional_0717/dose_statistics_results/histograms/dose_statistics_results/compiled_dose_statistics.csv', index=False)
```
